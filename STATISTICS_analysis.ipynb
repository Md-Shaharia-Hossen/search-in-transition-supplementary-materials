```python
# ============================================================
# STATISTICAL ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt


# ============================================================
# 1. LOAD DATA
# ============================================================

file_path = "Search in Transition_final versione.xlsx"

df = pd.read_excel(
    file_path,
    sheet_name="Form responses 1"
)


# ============================================================
# 2. CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
    .astype(str)
    .str.replace("\n", " ", regex=False)
    .str.strip()
)


# ============================================================
# 3. KEEP ONLY VALID PARTICIPANTS
# ============================================================

consent_col = (
    "I voluntarily agree to participate in this study."
)

df = df[
    df[consent_col]
    .astype(str)
    .str.strip()
    .eq("Yes, I agree")
].copy()


print("Valid participants:", len(df))


# ============================================================
# 4. COLUMN NAMES
# ============================================================

llm_frequency_col = (
    "How often do you use LLMs "
    "(e.g., ChatGPT) for English test preparation?"
)

search_frequency_col = (
    "How often do you use Google or other search engines "
    "for English test preparation?"
)

llm_satisfaction_col = (
    "How satisfied are you with the accuracy of LLMs "
    "in solving English test problems?"
)

search_satisfaction_col = (
    "How satisfied are you with the accuracy of search engines "
    "in solving English test problems?"
)

llm_efficiency_col = (
    "How efficient are LLMs in helping you complete "
    "English test tasks?"
)

search_efficiency_col = (
    "How efficient are search engines in helping you complete "
    "English test tasks?"
)

llm_ease_col = (
    "How easy is it to use LLMs for English test problem solving?"
)

search_ease_col = (
    "How easy is it to use search engines for English test problem solving?"
)


# ============================================================
# 5. LIKERT SCALE CODING
# ============================================================

frequency_mapping = {

    "Never": 1,

    "Occasionally": 2,

    "Sometimes": 3,

    "Often": 4,

    "Always": 5

}


satisfaction_mapping = {

    "Very Dissatisfied": 1,

    "Dissatisfied": 2,

    "Neutral": 3,

    "Satisfied": 4,

    "Very Satisfied": 5

}


efficiency_mapping = {

    "Not Efficient at All": 1,

    "Slightly Efficient": 2,

    "Moderately Efficient": 3,

    "Very Efficient": 4,

    "Extremely Efficient": 5

}


ease_mapping = {

    "Very Difficult": 1,

    "Difficult": 2,

    "Neutral": 3,

    "Easy": 4,

    "Very Easy": 5

}


# ============================================================
# 6. CONVERT CATEGORICAL RESPONSES TO NUMERIC SCORES
# ============================================================

df["LLM_Use_Frequency"] = (
    df[llm_frequency_col]
    .map(frequency_mapping)
)

df["Search_Use_Frequency"] = (
    df[search_frequency_col]
    .map(frequency_mapping)
)

df["LLM_Satisfaction"] = (
    df[llm_satisfaction_col]
    .map(satisfaction_mapping)
)

df["Search_Satisfaction"] = (
    df[search_satisfaction_col]
    .map(satisfaction_mapping)
)

df["LLM_Efficiency"] = (
    df[llm_efficiency_col]
    .map(efficiency_mapping)
)

df["Search_Efficiency"] = (
    df[search_efficiency_col]
    .map(efficiency_mapping)
)

df["LLM_Ease"] = (
    df[llm_ease_col]
    .map(ease_mapping)
)

df["Search_Ease"] = (
    df[search_ease_col]
    .map(ease_mapping)
)


# ============================================================
# 7. DESCRIPTIVE STATISTICS
# ============================================================

analysis_columns = [

    "LLM_Use_Frequency",

    "Search_Use_Frequency",

    "LLM_Satisfaction",

    "Search_Satisfaction",

    "LLM_Efficiency",

    "Search_Efficiency",

    "LLM_Ease",

    "Search_Ease"

]


descriptive_statistics = (

    df[analysis_columns]

    .describe()

    .T

)


descriptive_statistics[

    "95% CI Lower"

] = (

    descriptive_statistics["mean"]

    -

    1.96

    *

    descriptive_statistics["std"]

    /

    np.sqrt(descriptive_statistics["count"])

)


descriptive_statistics[

    "95% CI Upper"

] = (

    descriptive_statistics["mean"]

    +

    1.96

    *

    descriptive_statistics["std"]

    /

    np.sqrt(descriptive_statistics["count"])

)


print("\n====================================")
print("DESCRIPTIVE STATISTICS")
print("====================================")

print(

    descriptive_statistics[
        [
            "count",
            "mean",
            "std",
            "min",
            "50%",
            "max",
            "95% CI Lower",
            "95% CI Upper"
        ]
    ]

)


# Save results

descriptive_statistics.to_csv(

    "descriptive_statistics.csv"

)


# ============================================================
# 8. PAIRED COMPARISON: LLM VS SEARCH ENGINE
# ============================================================

paired_comparisons = {

    "Usage Frequency": (

        "LLM_Use_Frequency",

        "Search_Use_Frequency"

    ),

    "Satisfaction": (

        "LLM_Satisfaction",

        "Search_Satisfaction"

    ),

    "Efficiency": (

        "LLM_Efficiency",

        "Search_Efficiency"

    ),

    "Ease of Use": (

        "LLM_Ease",

        "Search_Ease"

    )

}


comparison_results = []


for measure, columns in paired_comparisons.items():

    llm_scores = df[columns[0]].dropna()

    search_scores = df[columns[1]].dropna()


    # Keep equal-length paired observations

    paired_data = pd.concat(

        [

            df[columns[0]],

            df[columns[1]]

        ],

        axis=1

    ).dropna()


    llm_scores = paired_data[columns[0]]

    search_scores = paired_data[columns[1]]


    # Paired t-test

    t_statistic, p_value = stats.ttest_rel(

        llm_scores,

        search_scores

    )


    comparison_results.append(

        {

            "Measure": measure,

            "LLM Mean": llm_scores.mean(),

            "Search Engine Mean": search_scores.mean(),

            "Mean Difference":

            llm_scores.mean()

            -

            search_scores.mean(),

            "t-statistic": t_statistic,

            "p-value": p_value

        }

    )


comparison_df = pd.DataFrame(

    comparison_results

)


print("\n====================================")
print("PAIRED COMPARISON RESULTS")
print("====================================")

print(comparison_df)


comparison_df.to_csv(

    "paired_comparison_results.csv",

    index=False

)


# ============================================================
# 9. ONE-WAY ANOVA
# ============================================================

# ANOVA across the four quantitative measures
# This is descriptive exploratory analysis.

anova_results = []


for tool, columns in {

    "LLM": [

        "LLM_Use_Frequency",

        "LLM_Satisfaction",

        "LLM_Efficiency",

        "LLM_Ease"

    ],

    "Search Engine": [

        "Search_Use_Frequency",

        "Search_Satisfaction",

        "Search_Efficiency",

        "Search_Ease"

    ]

}.items():


    groups = [

        df[column].dropna()

        for column in columns

    ]


    f_statistic, p_value = stats.f_oneway(

        *groups

    )


    anova_results.append(

        {

            "Tool": tool,

            "F-statistic": f_statistic,

            "p-value": p_value

        }

    )


anova_df = pd.DataFrame(

    anova_results

)


print("\n====================================")
print("ANOVA RESULTS")
print("====================================")

print(anova_df)


anova_df.to_csv(

    "anova_results.csv",

    index=False

)


# ============================================================
# 10. CHI-SQUARE TEST
# ============================================================

# Example:
# Gender × Overall Tool Preference

gender_col = "What is your gender?"

preference_col = (
    "Which tool do you prefer overall "
    "for English test preparation?"
)


contingency_table = pd.crosstab(

    df[gender_col],

    df[preference_col]

)


chi2, p_value, degrees_of_freedom, expected = (

    stats.chi2_contingency(

        contingency_table

    )

)


print("\n====================================")
print("CHI-SQUARE TEST")
print("====================================")

print(

    "Chi-square statistic:",

    chi2

)

print(

    "Degrees of freedom:",

    degrees_of_freedom

)

print(

    "p-value:",

    p_value

)


chi_square_results = pd.DataFrame(

    {

        "Chi-square": [chi2],

        "Degrees of Freedom": [

            degrees_of_freedom

        ],

        "p-value": [p_value]

    }

)


chi_square_results.to_csv(

    "chi_square_results.csv",

    index=False

)


# ============================================================
# 11. BOX PLOT
# ============================================================

plt.figure(

    figsize=(11, 7)

)


plt.boxplot(

    [

        df[column].dropna()

        for column in analysis_columns

    ],

    vert=False,

    labels=[

        "LLM Ease",

        "LLM Efficiency",

        "LLM Satisfaction",

        "LLM Use Frequency",

        "Search Ease",

        "Search Efficiency",

        "Search Satisfaction",

        "Search Use Frequency"

    ],

    showmeans=True

)


plt.title(

    "Distribution of Quantitative Survey Measures",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(

    "Likert Score (1–5)"

)


plt.xlim(

    0.8,

    5.2

)


plt.xticks(

    [1, 2, 3, 4, 5]

)


plt.grid(

    axis="x",

    linestyle="--",

    alpha=0.5

)


plt.tight_layout()


plt.savefig(

    "statistical_analysis_boxplot.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 12. SAVE COMPLETE ANALYSIS DATA
# ============================================================

df.to_csv(

    "complete_analysis_dataset.csv",

    index=False

)


print("\n====================================")
print("STATISTICAL ANALYSIS COMPLETED")
print("====================================")
```

```python
# ============================================================
# CORRECTED STATISTICAL ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt


# ============================================================
# 1. LOAD DATA
# ============================================================

file_path = "Search in Transition_final versione.xlsx"

df = pd.read_excel(
    file_path,
    sheet_name="Form responses 1"
)


# ============================================================
# 2. CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
    .astype(str)
    .str.replace("\n", " ", regex=False)
    .str.strip()
)


# ============================================================
# 3. KEEP VALID PARTICIPANTS ONLY
# ============================================================

consent_col = (
    "I voluntarily agree to participate in this study."
)

df = df[
    df[consent_col]
    .astype(str)
    .str.strip()
    .eq("Yes, I agree")
].copy()

print("Valid participants:", len(df))


# ============================================================
# 4. COLUMN DEFINITIONS
# ============================================================

llm_frequency_col = (
    "How often do you use LLMs "
    "(e.g., ChatGPT) for English test preparation?"
)

search_frequency_col = (
    "How often do you use Google or other search engines "
    "for English test preparation?"
)

llm_satisfaction_col = (
    "How satisfied are you with the accuracy of LLMs "
    "in solving English test problems?"
)

search_satisfaction_col = (
    "How satisfied are you with the accuracy of search engines "
    "in solving English test problems?"
)

llm_efficiency_col = (
    "How efficient are LLMs in helping you complete "
    "English test tasks?"
)

search_efficiency_col = (
    "How efficient are search engines in helping you complete "
    "English test tasks?"
)

llm_ease_col = (
    "How easy is it to use LLMs for English test problem solving?"
)

search_ease_col = (
    "How easy is it to use search engines for English test problem solving?"
)


# ============================================================
# 5. CORRECT LIKERT CODING
# ============================================================

frequency_mapping = {

    "Never": 1,

    "Occasionally": 2,

    "Sometimes": 3,

    "Often": 4,

    "Always": 5

}


satisfaction_mapping = {

    "Very Dissatisfied": 1,

    "Dissatisfied": 2,

    "Neutral": 3,

    "Satisfied": 4,

    "Very Satisfied": 5

}


efficiency_mapping = {

    "Not Efficient at All": 1,

    "Slightly Efficient": 2,

    "Moderately Efficient": 3,

    "Very Efficient": 4,

    "Extremely Efficient": 5

}


ease_mapping = {

    "Very Difficult": 1,

    "Difficult": 2,

    "Neutral": 3,

    "Easy": 4,

    "Very Easy": 5

}


# ============================================================
# 6. CONVERT RESPONSES TO NUMERIC SCORES
# ============================================================

df["LLM_Use_Frequency"] = (
    df[llm_frequency_col]
    .map(frequency_mapping)
)

df["Search_Use_Frequency"] = (
    df[search_frequency_col]
    .map(frequency_mapping)
)

df["LLM_Satisfaction"] = (
    df[llm_satisfaction_col]
    .map(satisfaction_mapping)
)

df["Search_Satisfaction"] = (
    df[search_satisfaction_col]
    .map(satisfaction_mapping)
)

df["LLM_Efficiency"] = (
    df[llm_efficiency_col]
    .map(efficiency_mapping)
)

df["Search_Efficiency"] = (
    df[search_efficiency_col]
    .map(efficiency_mapping)
)

df["LLM_Ease"] = (
    df[llm_ease_col]
    .map(ease_mapping)
)

df["Search_Ease"] = (
    df[search_ease_col]
    .map(ease_mapping)
)


# ============================================================
# 7. ANALYSIS VARIABLES
# ============================================================

analysis_columns = [

    "LLM_Use_Frequency",

    "Search_Use_Frequency",

    "LLM_Satisfaction",

    "Search_Satisfaction",

    "LLM_Efficiency",

    "Search_Efficiency",

    "LLM_Ease",

    "Search_Ease"

]


# ============================================================
# 8. DESCRIPTIVE STATISTICS
# ============================================================

descriptive_results = []


for column in analysis_columns:

    values = df[column].dropna()

    n = len(values)

    mean = values.mean()

    std = values.std(ddof=1)

    median = values.median()

    q1 = values.quantile(0.25)

    q3 = values.quantile(0.75)

    minimum = values.min()

    maximum = values.max()

    standard_error = std / np.sqrt(n)

    ci_lower = mean - 1.96 * standard_error

    ci_upper = mean + 1.96 * standard_error


    descriptive_results.append({

        "Variable": column,

        "N": n,

        "Mean": mean,

        "SD": std,

        "Median": median,

        "Q1": q1,

        "Q3": q3,

        "IQR": q3 - q1,

        "Minimum": minimum,

        "Maximum": maximum,

        "95% CI Lower": ci_lower,

        "95% CI Upper": ci_upper

    })


descriptive_df = pd.DataFrame(

    descriptive_results

)


print("\n====================================")
print("DESCRIPTIVE STATISTICS")
print("====================================")

print(

    descriptive_df.round(3)

)


descriptive_df.to_csv(

    "descriptive_statistics_corrected.csv",

    index=False

)


# ============================================================
# 9. WILCOXON SIGNED-RANK TEST
# ============================================================

# Used because:
# 1. The observations are paired
# 2. The same participants evaluated both tools
# 3. Likert responses are ordinal

paired_comparisons = {

    "Usage Frequency": (

        "LLM_Use_Frequency",

        "Search_Use_Frequency"

    ),

    "Satisfaction": (

        "LLM_Satisfaction",

        "Search_Satisfaction"

    ),

    "Efficiency": (

        "LLM_Efficiency",

        "Search_Efficiency"

    ),

    "Ease of Use": (

        "LLM_Ease",

        "Search_Ease"

    )

}


wilcoxon_results = []


for measure, columns in paired_comparisons.items():

    paired_data = df[

        [

            columns[0],

            columns[1]

        ]

    ].dropna()


    llm_scores = paired_data[columns[0]]

    search_scores = paired_data[columns[1]]


    # Wilcoxon signed-rank test

    statistic, p_value = stats.wilcoxon(

        llm_scores,

        search_scores,

        zero_method="wilcox",

        alternative="two-sided"

    )


    # Rank-biserial effect size

    differences = llm_scores - search_scores

    non_zero = differences[differences != 0]

    n_nonzero = len(non_zero)


    if n_nonzero > 0:

        ranks = stats.rankdata(

            abs(non_zero)

        )

        positive_rank_sum = ranks[

            non_zero > 0

        ].sum()


        negative_rank_sum = ranks[

            non_zero < 0

        ].sum()


        rank_biserial = (

            positive_rank_sum

            -

            negative_rank_sum

        ) / (

            positive_rank_sum

            +

            negative_rank_sum

        )

    else:

        rank_biserial = np.nan


    wilcoxon_results.append({

        "Measure": measure,

        "LLM Median": llm_scores.median(),

        "Search Median": search_scores.median(),

        "LLM Mean": llm_scores.mean(),

        "Search Mean": search_scores.mean(),

        "Wilcoxon W": statistic,

        "p-value": p_value,

        "Rank-biserial Effect Size": rank_biserial

    })


wilcoxon_df = pd.DataFrame(

    wilcoxon_results

)


print("\n====================================")
print("WILCOXON SIGNED-RANK TEST")
print("====================================")

print(

    wilcoxon_df.round(4)

)


wilcoxon_df.to_csv(

    "wilcoxon_paired_comparison.csv",

    index=False

)


# ============================================================
# 10. FRIEDMAN TEST
# ============================================================

# The Friedman test is used for repeated ordinal measurements
# from the same participants.

friedman_results = []


friedman_groups = {

    "LLM": [

        "LLM_Use_Frequency",

        "LLM_Satisfaction",

        "LLM_Efficiency",

        "LLM_Ease"

    ],

    "Search Engine": [

        "Search_Use_Frequency",

        "Search_Satisfaction",

        "Search_Efficiency",

        "Search_Ease"

    ]

}


for tool, columns in friedman_groups.items():

    repeated_data = df[

        columns

    ].dropna()


    statistic, p_value = stats.friedmanchisquare(

        *[

            repeated_data[column]

            for column in columns

        ]

    )


    n = len(repeated_data)

    k = len(columns)


    # Kendall's W effect size

    kendalls_w = statistic / (

        n * (k - 1)

    )


    friedman_results.append({

        "Tool": tool,

        "Friedman Chi-square": statistic,

        "df": k - 1,

        "p-value": p_value,

        "Kendall's W": kendalls_w

    })


friedman_df = pd.DataFrame(

    friedman_results

)


print("\n====================================")
print("FRIEDMAN TEST")
print("====================================")

print(

    friedman_df.round(4)

)


friedman_df.to_csv(

    "friedman_test_results.csv",

    index=False

)


# ============================================================
# 11. CHI-SQUARE TEST OF INDEPENDENCE
# ============================================================

gender_col = (

    "What is your gender?"

)


preference_col = (

    "Which tool do you prefer overall "

    "for English test preparation?"

)


contingency_table = pd.crosstab(

    df[gender_col],

    df[preference_col]

)


chi2, p_value, degrees_of_freedom, expected = (

    stats.chi2_contingency(

        contingency_table

    )

)


# Cramer's V effect size

n = contingency_table.to_numpy().sum()

r, c = contingency_table.shape

cramers_v = np.sqrt(

    chi2 /

    (

        n *

        min(

            r - 1,

            c - 1

        )

    )

)


print("\n====================================")
print("CHI-SQUARE TEST OF INDEPENDENCE")
print("====================================")

print(

    "Chi-square statistic:",

    round(chi2, 4)

)

print(

    "Degrees of freedom:",

    degrees_of_freedom

)

print(

    "p-value:",

    round(p_value, 4)

)

print(

    "Cramer's V:",

    round(cramers_v, 4)

)


chi_square_df = pd.DataFrame({

    "Chi-square": [chi2],

    "Degrees of Freedom": [

        degrees_of_freedom

    ],

    "p-value": [p_value],

    "Cramer's V": [cramers_v]

})


chi_square_df.to_csv(

    "chi_square_results_corrected.csv",

    index=False

)


# ============================================================
# 12. BOX PLOT USING ACTUAL SURVEY DATA
# ============================================================

plot_columns = [

    "LLM_Ease",

    "LLM_Efficiency",

    "LLM_Satisfaction",

    "LLM_Use_Frequency",

    "Search_Ease",

    "Search_Efficiency",

    "Search_Satisfaction",

    "Search_Use_Frequency"

]


plot_labels = [

    "LLM Ease",

    "LLM Efficiency",

    "LLM Satisfaction",

    "LLM Use Frequency",

    "Search Ease",

    "Search Efficiency",

    "Search Satisfaction",

    "Search Use Frequency"

]


plt.figure(

    figsize=(11, 7)

)


plt.boxplot(

    [

        df[column].dropna()

        for column in plot_columns

    ],

    vert=False,

    labels=plot_labels,

    showmeans=True

)


plt.title(

    "Distribution of Survey Measures for LLMs and Search Engines",

    fontsize=14,

    fontweight="bold"

)


plt.xlabel(

    "Likert Score (1 = Lowest, 5 = Highest)"

)


plt.xlim(

    0.8,

    5.2

)


plt.xticks(

    [1, 2, 3, 4, 5]

)


plt.grid(

    axis="x",

    linestyle="--",

    alpha=0.5

)


plt.tight_layout()


plt.savefig(

    "corrected_boxplot_actual_data.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 13. SAVE ANALYSIS DATA
# ============================================================

df.to_csv(

    "cleaned_analysis_dataset.csv",

    index=False

)


# ============================================================
# 14. FINAL SUMMARY
# ============================================================

print("\n====================================")
print("CORRECTED STATISTICAL ANALYSIS COMPLETE")
print("====================================")

print("Valid sample size:", len(df))

print("\nGenerated files:")

print("1. descriptive_statistics_corrected.csv")

print("2. wilcoxon_paired_comparison.csv")

print("3. friedman_test_results.csv")

print("4. chi_square_results_corrected.csv")

print("5. corrected_boxplot_actual_data.png")

print("6. cleaned_analysis_dataset.csv")
```
